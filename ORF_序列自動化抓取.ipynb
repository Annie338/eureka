```python
def transcription(directory):#(directory)是這個函式接受的一個參數
  #f = open(directory, "r")
  #"r"是模式字串，以「讀取」(read) 模式開啟檔案，但不能修改或寫入檔案
  Tr = list()
  Tr1 = list()
  for line in directory:
    if list(line)[0] == ">":
      continue #[0]是索引可讀取第一個字元
    else:
      for i in list(line):
        if i == "A":
          Tr.append("U")
          Tr1.append("A")
          #append「追加」一個序列（特別是列表 list）的末尾添加一個新的元素。
        elif i == "T":
          Tr1.append("U")
          Tr.append("A")
        elif i == "C":
          Tr1.append("C")
          Tr.append("G")
        elif i == "G":
          Tr1.append("G")
          Tr.append("C")
  Tr =Tr[::-1] #將序列反轉
  result = ["".join(Tr), "".join(Tr1)]
  return result #連接(join)字串元素時，元素之間沒有分隔符號，作為函式的結果返回
```

```python
# 標準遺傳密碼表(RNA密碼子到氨基酸)
#字典 (Dictionary):是一系列無序的key-值，可儲存具有關聯性的資料，key是唯一
rnaProtein = {
    'UUU': 'F', 'UUC': 'F', 'UUA': 'L', 'UUG': 'L',
    'CUU': 'L', 'CUC': 'L', 'CUA': 'L', 'CUG': 'L',
    'AUU': 'I', 'AUC': 'I', 'AUA': 'I', 'AUG': 'M',  # AUG is also the start codon
    'GUU': 'V', 'GUC': 'V', 'GUA': 'V', 'GUG': 'V',

    'UCU': 'S', 'UCC': 'S', 'UCA': 'S', 'UCG': 'S',
    'CCU': 'P', 'CCC': 'P', 'CCA': 'P', 'CCG': 'P',
    'ACU': 'T', 'ACC': 'T', 'ACA': 'T', 'ACG': 'T',
    'GCU': 'A', 'GCC': 'A', 'GCA': 'A', 'GCG': 'A',

    'UAU': 'Y', 'UAC': 'Y', 'UAA': '*',  # * indicates a stop codon
    'CAU': 'H', 'CAC': 'H', 'CAA': 'Q', 'CAG': 'Q',
    'AAU': 'N', 'AAC': 'N', 'AAA': 'K', 'AAG': 'K',
    'GAU': 'D', 'GAC': 'D', 'GAA': 'E', 'GAG': 'E',

    'UGU': 'C', 'UGC': 'C', 'UGA': '*',  # * indicates a stop codon
    'CGU': 'R', 'CGC': 'R', 'CGA': 'R', 'CGG': 'R',
    'AGU': 'S', 'AGC': 'S', 'AGA': 'R', 'AGG': 'R',
    'GGU': 'G', 'GGC': 'G', 'GGA': 'G', 'GGG': 'G',

    'UGG': 'W', 'UAG': '*'   # * indicates a stop codon
}
```

```python
def translation(directory):
  forwardSeq = []
  fs = directory[0]
  for n in range(0,3):
    strSeq = [] #儲存翻譯後的氨基酸
    for i in range(n, len(fs), 3):
      codon = fs[i:i+3] #len(fs)是序列的長度
      if len(codon) == 3: #三個核苷酸，形成一個密碼子codon
        strSeq.append(rnaProtein[codon])
    forwardSeq.append("".join(strSeq))
    #查字典丟進去strSeq

#反向
  reversedSeq = []
  rs = directory[1]
  for m in range(0,3):
    revSeq = []
    for l in range(m, len(rs), 3):
      codon1 = rs[l:l+3]
      if len(codon1) == 3:
        revSeq.append(rnaProtein[codon1])
    reversedSeq.append("".join(revSeq))

  return forwardSeq, reversedSeq
```

```python
def screenSeq(directory):
  forwardSeq = directory[0]
  reversedSeq = directory[1]
  result = []
  #正向
  for f in range(0,3): #range(0,3)是因為正/反股各有三條
    mIndexes = [i for i, value in enumerate(list(forwardSeq[f])) if value == 'M']
    endIndexes = [i for i, value in enumerate(list(forwardSeq[f])) if value == '*']
    #得到M跟*的本人和位置
    for n in mIndexes:
      for k in endIndexes:
        if n < k:
          result.append(forwardSeq[f][n:k+1])
          # +1確保包含k本人的*
          break #確保是最近的那個，用return會重跑覆蓋

  #反向
  for r in range(0,3):
    mIndexes = [i for i, value in enumerate(list(reversedSeq[r])) if value == 'M']
    endIndexes = [i for i, value in enumerate(list(reversedSeq[r])) if value == '*']
    for m in mIndexes:
      for t in endIndexes:
        if m < t:
          result.append(reversedSeq[r][m:t+1])
          break

  print(result)
  longResult = "" #儲存的是字母，用字串
  if result: #檢查列表是否為空
    longResult = max(result, key=len)
    #max對result的每個元素應用len(用於獲取序列的長度)，根據回傳的key比大小


  return (longResult)
```

## 提取 m 和 * 之间的序列 (考虑最近的*)

###
找到所有 'M' 的索引。對於每個 'M' 的索引，找到後面最近的 '*' 的索引，提取以該 'M' 開頭並以該 '*' 结尾的子序列。


```python
def extract_orf_sequences(translated_sequences):
    extracted_sequences = []

    for frame_sequences in translated_sequences:
        for seq in frame_sequences:
            m_indexes = [i for i, value in enumerate(seq) if value == 'M']
            end_indexes = [i for i, value in enumerate(seq) if value == '*']

            for m_index in m_indexes:
                # Find the first '*' after the current 'M'
                for end_index in end_indexes:
                    if end_index > m_index:
                        # Extract the subsequence and break the loop
                        # Include the stop codon '*' in the extracted sequence
                        extracted_sequences.append(seq[m_index : end_index + 1])
                        break # This break ensures we get the nearest '*'

    # 在提取完所有序列後，尋找最長的序列
    longest_sequence = ""
    if extracted_sequences: # 檢查列表是否為空，避免在空列表上比較
        longest_sequence = max(extracted_sequences, key=len)


    return longest_sequence # 返回最長的序列
```
